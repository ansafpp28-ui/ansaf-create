```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
sns.set_theme(style="whitegrid")
```

```python
df = pd.read_csv("/content/county_40017_loan_purposes_1-2-4-5-31-32.csv", low_memory=False)
```

```python
df
```

```python
df.columns = [
    str(col).strip().lower().replace(" ", "_")
    for col in df.columns
]
```

```python
print("Dataset shape:", df.shape)
print("\nFirst 5 rows:")
display(df.head())
```

```python
df["action_taken"] = pd.to_numeric(
    df["action_taken"], errors="coerce"
)

eda_df = df[df["action_taken"].isin([1, 3])].copy()

eda_df["loan_approved"] = (
    eda_df["action_taken"] == 1
).astype(int)

eda_df["outcome"] = eda_df["loan_approved"].map({
    0: "Denied",
    1: "Originated"
})

print("\nEDA dataset shape:", eda_df.shape)
print("\nTarget distribution:")
print(eda_df["outcome"].value_counts())
```

```python
plot_df = eda_df.sample(
    n=min(5000, len(eda_df)),
    random_state=42
)
```

```python
plt.figure(figsize=(7, 5))
sns.countplot(data=plot_df, x="outcome")
plt.title("Loan Application Outcome Distribution")
plt.xlabel("Application Outcome")
plt.ylabel("Number of Applications")
plt.tight_layout()
plt.show()
```

```python
if "loan_amount" in plot_df.columns:
    loan_amount = pd.to_numeric(
        plot_df["loan_amount"], errors="coerce"
    ).dropna()

    plt.figure(figsize=(8, 5))
    sns.histplot(loan_amount, bins=40, kde=True)
    plt.title("Distribution of Loan Amount")
    plt.xlabel("Loan Amount")
    plt.ylabel("Frequency")
    plt.tight_layout()
    plt.show()
```

```python
if "income" in plot_df.columns:
    income = pd.to_numeric(
        plot_df["income"], errors="coerce"
    ).dropna()

    plt.figure(figsize=(8, 5))
    sns.histplot(income, bins=40, kde=True)
    plt.title("Distribution of Applicant Income")
    plt.xlabel("Income")
    plt.ylabel("Frequency")
    plt.tight_layout()
    plt.show()
```

```python
if "loan_amount" in plot_df.columns:
    box_df = plot_df.copy()
    box_df["loan_amount"] = pd.to_numeric(
        box_df["loan_amount"], errors="coerce"
    )
    box_df = box_df.dropna(subset=["loan_amount"])

    # Limit extreme values for a readable visualization
    upper = box_df["loan_amount"].quantile(0.99)
    box_df = box_df[box_df["loan_amount"] <= upper]

    plt.figure(figsize=(8, 5))
    sns.boxplot(
        data=box_df,
        x="outcome",
        y="loan_amount"
    )
    plt.title("Loan Amount by Application Outcome")
    plt.xlabel("Application Outcome")
    plt.ylabel("Loan Amount (up to 99th percentile)")
    plt.tight_layout()
    plt.show()
```

```python
numeric_df = plot_df.select_dtypes(
    include=np.number
).copy()

# Remove identifier-like columns from this visualization
numeric_df = numeric_df.drop(
    columns=[
        c for c in [
            "activity_year", "sequence_number",
            "loan_approved", "action_taken"
        ] if c in numeric_df.columns
    ],
    errors="ignore"
)

# Keep numeric features with sufficient observed data
numeric_df = numeric_df.loc[
    :, numeric_df.notna().sum() >= 20
]

# Limit the number of features for readability
numeric_df = numeric_df.iloc[:, :12]

if numeric_df.shape[1] >= 2:
    corr = numeric_df.corr()

    plt.figure(figsize=(12, 8))
    sns.heatmap(
        corr,
        annot=True,
        cmap="coolwarm",
        fmt=".2f",
        center=0
    )
    plt.title("Correlation Heatmap")
    plt.tight_layout()
    plt.show()
```

```python
if {"income", "loan_amount"}.issubset(plot_df.columns):
    scatter_df = plot_df[
        ["income", "loan_amount", "outcome"]
    ].copy()

    scatter_df["income"] = pd.to_numeric(
        scatter_df["income"], errors="coerce"
    )
    scatter_df["loan_amount"] = pd.to_numeric(
        scatter_df["loan_amount"], errors="coerce"
    )
    scatter_df = scatter_df.dropna()

    # Exclude extreme values for readability
    scatter_df = scatter_df[
        (scatter_df["income"] <= scatter_df["income"].quantile(0.99)) &
        (scatter_df["loan_amount"] <= scatter_df["loan_amount"].quantile(0.99))
    ]

    plt.figure(figsize=(8, 5))
    sns.scatterplot(
        data=scatter_df,
        x="income",
        y="loan_amount",
        hue="outcome",
        alpha=0.6
    )
    plt.title("Applicant Income vs Loan Amount")
    plt.xlabel("Applicant Income")
    plt.ylabel("Loan Amount")
    plt.tight_layout()
    plt.show()
```

```python
pair_cols = [
    c for c in [
        "loan_amount", "income",
        "property_value", "loan_to_value_ratio"
    ] if c in plot_df.columns
]

if len(pair_cols) >= 2:
    pair_df = plot_df[pair_cols + ["outcome"]].copy()

    for col in pair_cols:
        pair_df[col] = pd.to_numeric(
            pair_df[col], errors="coerce"
        )

    pair_df = pair_df.dropna()

    # Pair plots can be slow, so use a smaller sample
    pair_df = pair_df.sample(
        n=min(500, len(pair_df)),
        random_state=42
    )

    if len(pair_df) > 1:
        sns.pairplot(
            pair_df,
            vars=pair_cols,
            hue="outcome",
            diag_kind="hist"
        )
        plt.show()
```
